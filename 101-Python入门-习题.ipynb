# 函数

将上面的Fibonacci sequence计算过程封装为函数。

输入：n。`0<n<10**5`

输出：fib(n)。

```python
def fib(n):
    a, b = 1, 1
    while n>2:
        a, b, n = a+b, a, n-1
    return a

print(fib(10**2))
%time fib(30)
```

用递归实现Fibonacci sequence，并比较效率。

```python
def fib(n):
    if n <= 2:
        return 1
    return fib(n-1) + fib(n-2)

%time fib(30)
```

将上面Euclidean algorithm计算过程封装为函数。

输入：a, b，两者均满足`0<a<10**10`。并且a>=b。

输出：gcd(a, b)。

```python
def gcd(a, b):
    while b != 0:
        a, b = b, a%b
    return a

print(gcd(1071, 462))
%timeit gcd(1071, 462)
```

将Euclidean algorithm计算过程改为递归，并计算效率。

```python
def gcd(a, b):
    if b == 0:
        return a
    return gcd(b, a%b)

print(gcd(1071, 462))
%timeit gcd(1071, 462)
```

[Extended Euclidean algorithm](https://en.wikipedia.org/wiki/Extended_Euclidean_algorithm)是Euclidean algorithm的一个扩展，能够求出a, b最大公约数的同时，找到整数x, y，使其满足[Bézout's identity](https://en.wikipedia.org/wiki/B%C3%A9zout%27s_identity)。

    a*x+b*y=gcd(a,b)
    
其基本思路如下：

对于一般Euclidean algorithm，在求解gcd(a, b)时，倒数第三步应为`r(n-2)=r(n-1)*q(n)+gcd(a,b)`(见下表)。通过变形，这个式子可以变为`gcd(a,b)=r(n-2)-r(n-1)*q(n)`。注意这个式子具有`gcd(a,b)=Y*r(n-1)+X*r(n-2)`的特征。此处我们令`Y(n-1)=-q(n)`，`X(n-1)=1`，式子可表为`gcd(a,b)=Y(n-1)*r(n-1)+X(n-1)*r(n-2)`。

而倒数第四步的式子应为`r(n-3)=r(n-2)*q(n-1)+r(n-1)`，变形可得`r(n-1)=r(n-3)-r(n-2)*q(n-1)`。代入前式可得

    gcd(a,b)=Y(n-1)*r(n-1)+X(n-1)*r(n-2)
            =Y(n-1)*(r(n-3)-r(n-2)*q(n-1))+X(n-1)*r(n-2)
            =(X(n-1)-Y(n-1)*q(n-1))*r(n-2)+Y(n-1)*r(n-3)

此时我们令`Y(n-2)=X(n-1)-Y(n-1)*q(n-1)`，`X(n-2)=Y(n-1)`，这个式子同样具有`gcd(a,b)=Y*r(n-2)+X*r(n-3)`的特征。

依此类推，可以得到`gcd=Y(0)r(0)+X(0)r(-1)`的式子。而我们知道，`r(0)=b`, `r(-1)=a`。因此Y(0)和X(0)即为Bézout's identity的一个特解。

推算补全所有数字的话，可以得到计算表如下：

                                a        b      Y                    X
    a=b*q1+r1                   a        b      X(1)-Y(1)*q(1)       Y(1)
    b=r1*q2+r2                  b        r1     X(2)-Y(2)*q(2)       Y(2)
    ...
    r(n-3)=r(n-2)*q(n-1)+r(n-1) r(n-3)   r(n-2) X(n-1)-Y(n-1)*q(n-1) Y(n-1)
    r(n-2)=r(n-1)*q(n)+r(n)     r(n-2)   r(n-1) 0-1*q(n)             1
    r(n-1)=r(n)*q(n)+r(n+1)     r(n-1)   r(n)   1                    0
                                gcd(a,b) 0      0                    1

请实现扩展欧几里得算法，求出方程`7x+4y=1`的一个解。并做验算。

输入：a, b。均满足a>0，且可以假定a>b。

输出：x, y。满足`a*x+b*y=gcd(a,b)`

```python
# 来源[扩展欧几里得算法](https://zh.wikipedia.org/wiki/%E6%89%A9%E5%B1%95%E6%AC%A7%E5%87%A0%E9%87%8C%E5%BE%97%E7%AE%97%E6%B3%95)
def ext_euclid(a, b):
    if b == 0:
        return 1, 0, a
    x, y, q = ext_euclid(b, a % b)
    return y, (x - (a // b) * y), q

x, y, q = ext_euclid(7, 4)
print(x, y, q)
print(7*x+4*y)
```

对于一次不定方程`ax+by=c`，如x0,y0为其一个解，则方程的一切解都具有如下的形式。

    x = x0-(b/gcd(a,b))*t
    y = y0+(a/gcd(a,b))*t
    
其中t为整数。

基于以上知识，给出百鸡问题的所有正整数解组，并且验算是否满足原始的百鸡问题。

```python
# 首先需要解出系数a1, b1。实际由于可以直接看出gcd(7,4)=1，因此a1=7, b1=4

# 上文给出了Bézout's identity：7x+4y=1的一个特解，x=-1，y=2。我们又知道百鸡问题满足可解条件100 mod gcd(7,4) = 0。
# 因此我们可以向两边乘以系数s = 100/gcd(7,4) = 100。得到7x+4y=100的一个特解，x0=-100，y0=200。

# 同时，解需要满足正数要求，因此
# -100-4*t>=0         t<=-25
# 200+7*t>=0          t>=-200/7=-28.57>-29

for t in range(-28,-24):
    x, y = -100-4*t, 200+7*t
    z = 300 - 15*x - 9*y
    print(x, y, z)
    # 验算
    if 5*x + 3*y + int(z/3) != 100:
        print('error')
    if x + y + z != 100:
        print('error')
```

将百鸡问题推广到一般二元一次不定方程，形如`a1*x1+a2*x2=c`。判定其可解性，并求出`x1>=0`且`x2>=0`范围内所有可能解。

输入：a1, a2, c。皆为正整数，取值范围`0<c<2**31-1`。

输出：打印出所有可能解。

tips: 可以使用上文已经封装的所有函数。

```python
# 限定方程组为
# X1 + q1*t >= 0
# X2 + q2*t >= 0
# 并且q1 < 0, q2 > 0
def solve(a1, a2, c):
    X1, X2, g = ext_euclid(a1, a2)
    if c % g != 0:
        print('no result')
        return
    X1, X2 = X1*c/g, X2*c/g
    print('one match: %d %d' % (X1, X2))
    q1, q2 = -int(a2/g), int(a1/g)
    for t in range(-int(X2/q2), -int(X1/q1)+1):
        x1, x2 = X1 + q1*t, X2 + q2*t
        print('rslt:', t, int(x1), int(x2))

solve(7, 4, 100)
```
